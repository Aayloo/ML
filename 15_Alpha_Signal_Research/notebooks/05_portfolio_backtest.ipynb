# 05 · Portfolio backtest

Translate signal ranks into a neutral long-short portfolio and subtract turnover-based transaction costs.

```python
from src.data import make_demo_market_data
from src.features import build_features
from src.signals import build_signal
from src.portfolio import rank_portfolio, backtest

features = build_features(make_demo_market_data(), {'horizon': 5})
signal = build_signal(features, 'momentum')
weights = rank_portfolio(signal, n_quantiles=5, long_short=True)
returns = features[['date', 'asset']].copy()
returns['asset_return'] = features.groupby('asset')['close'].pct_change()
portfolio = backtest(weights, returns, cost_bps=5.0)
portfolio.tail()
```
