# 03 · Factor analysis

Measure predictive power before constructing a portfolio: daily Spearman IC and top-minus-bottom quantile spread.

```python
from src.data import make_demo_market_data
from src.features import build_features
from src.signals import build_signal
from src.metrics import cross_sectional_ic, quantile_spread

features = build_features(make_demo_market_data(), {'horizon': 5})
signal = build_signal(features, 'momentum')
analysis = signal.merge(features[['date', 'asset', 'forward_return']], on=['date', 'asset'])
ic = cross_sectional_ic(analysis)
spread = quantile_spread(analysis)
ic.describe(), spread.describe()
```
