# Machine Learning Model Learning Roadmap

This roadmap helps you quickly locate your position in the course. The main thread is "**Basics -> Classic Models -> Unsupervised Learning -> Ensembles & Advanced Topics -> Engineering**": the first two chapters build a complete regression/classification loop with linear models; chapters 03-06 cover four classic classification models; chapter 07 introduces unsupervised clustering; chapter 08 wraps up five single models with a credit-card default case; chapter 09 pushes performance further with ensemble learning (Random Forest / GBDT / XGBoost), followed by two special topics - PCA for dimensionality reduction and kernel methods - and finally feature engineering plus a comprehensive project before backtesting. This series focuses on "theory + cases"; engineering concerns such as model deployment are planned for future work.

**Data conventions (reproducible across machines)**: the main line uses only datasets built into sklearn or simulated data with fixed random seeds - zero downloads, zero license risk, and it runs on any machine with the standard libraries (numpy / matplotlib / scikit-learn / xgboost). There are five exceptions: the credit-card default data used in the comprehensive case (08, 09) and the engineering case (12) comes from a local CSV copy of the official UCI dataset (CC BY 4.0); chapter 13 uses the iris dataset from the official seaborn data repository on GitHub (iris.csv); chapter 14 uses Bitcoin data from the official Coin Metrics repository on GitHub (btc.csv); chapter 15 uses the clothing-image data from the official Fashion-MNIST repository on GitHub; chapter 16 uses daily US equity sample data from the official QuantConnect/Lean repository on GitHub; chapter 17 uses the SMS spam data from the justmarkham/pycon-2016-tutorial repository (a cleaned version of the UCI SMS Spam Collection). All of them are stored as local copies alongside the notebooks, so everything remains reproducible offline.

## Recommended Learning Order

1. **Basics (01-02)**: linear regression builds regression evaluation and gradient descent; logistic regression extends linear models to probabilistic classification - both share the same "loss -> gradient -> update" paradigm.
2. **Classic classification (03-06)**: decision trees, naive Bayes, SVM, and KNN represent four different inductive biases: partition rules, probabilistic inference, margin maximization, and local memory.
3. **Unsupervised & comprehensive (07-08)**: K-Means groups unlabeled data; the credit-card case puts five single models into a unified evaluation framework, comparing cross-validation, thresholds, calibration, and feature importance.
4. **Ensembles (09)**: Random Forest (Bagging) and GBDT/XGBoost (Boosting) are the workhorses for tabular data - note the difference between parallel variance reduction and sequential bias reduction.
5. **Advanced topics (10-11)**: PCA handles high-dimensional redundancy, speed, and visualization; kernel methods generalize the SVM kernel trick from chapter 05 to ridge regression and PCA.
6. **Engineering & projects (12-17)**: feature engineering and tuning complete the engineering chain; chapter 13 benchmarks 16 models (3 unsupervised + 9 classification + 4 regression) on iris under one scale; chapter 14 applies the skeleton to Bitcoin time series, with regression as the main thread and a "return vs volatility" comparison; chapter 15 applies it to Fashion-MNIST images, covering high-dimensional pixel handling and error analysis; chapter 16 completes a quant + ML project (cross-sectional factors -> all-model benchmark -> backtesting loop); chapter 17 completes text classification (TF-IDF -> all-model benchmark -> spam-word diagnosis). Together they cover five data modalities: tabular / time series / image / finance / text. Further quant extensions (stock pools, deeper factor research, portfolio optimization) can be continued in a separate quant research project.

## Three Chapter Templates

**Template A: model chapters (01-11)** - model theory -> NumPy implementation from scratch -> sklearn comparison & real cases -> chapter summary with validation results.

**Template B: engineering chapter (12)** - workflow overview -> step-by-step demo -> end-to-end walkthrough -> summary; no mathematical derivations or from-scratch implementations, staying close to real engineering practice.

**Template C: project chapters (13-17)** - shared skeleton: **data -> exploration (unsupervised first) -> feature engineering -> all-model benchmark -> diagnosis -> conclusion**, adapted to each data modality:

| Project chapter | Exploration | Feature Engineering | Diagnosis |
|---|---|---|---|
| 13 tabular | PCA / kernel PCA / K-Means | standardization + derived features | decision boundaries / confusion matrix |
| 14 time series | volatility clustering | lags + rolling features | residuals + comparison (return vs volatility) |
| 15 image | PCA projection / image reconstruction | standardization + PCA reduction | error samples + reduction comparison |
| 16 financial panel | K-Means market states / PCA market factors | cross-sectional factors + cross-sectional standardization | **backtest**: equity curve / drawdown / monthly heatmap |
| 17 text | word frequency / word cloud | TF-IDF (unigrams + bigrams) | error samples / spam-word diagnosis |

**Standard case deliverables**: an independent test set (or out-of-sample validation), a metrics table, and visualizations. Classification focuses on confusion matrices and ROC; regression on true-vs-predicted values, residuals, and R-squared; clustering on cluster structure and parameter selection (elbow method, silhouette score); ensembles and tree models on feature importance; dimensionality reduction on explained variance, reconstruction error, and downstream performance; quant on IC, AUC, equity curve, and drawdown.

## Learning & Implementation References

- Zhou Zhihua, *Machine Learning* (the "Watermelon Book"): main textbook - Ch. 3 "Linear Models", Ch. 4 "Decision Trees", Ch. 6 "Support Vector Machines", Ch. 8 "Ensemble Learning", Ch. 10 "Dimensionality Reduction and Metric Learning".
- [Datawhale *Pumpkin Book*](https://github.com/datawhalechina/pumpkin-book): companion reference for derivations.
- [hands-on-ML](https://github.com/ForceInjection/hands-on-ML): runnable notebooks aligned chapter by chapter.
- [Feature Engineering for Machine Learning](https://www.oreilly.com/library/view/feature-engineering-for/9781491953235/): the most authoritative dedicated reference on feature engineering.
- [Google Machine Learning Crash Course: Feature Engineering](https://developers.google.com/machine-learning/crash-course/representation/feature-engineering): official introductory framework.
- [scikit-learn documentation](https://scikit-learn.org/stable/): see [Preprocessing](https://scikit-learn.org/stable/modules/preprocessing.html), [Cross-validation](https://scikit-learn.org/stable/modules/cross_validation.html), and [Feature selection](https://scikit-learn.org/stable/modules/feature_selection.html).
- [XGBoost documentation](https://xgboost.readthedocs.io/): official reference for XGBoost in chapter 09.
