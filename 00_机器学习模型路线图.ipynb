# 机器学习模型学习路线图

这份路线图用于快速定位课程进度。课程主线为“**基础 → 经典模型 → 无监督 → 集成与进阶 → 工程**”：前两章用线性模型建立回归与分类闭环，03–06 覆盖四种经典分类模型，07 进入无监督聚类，08 用信用卡违约综合案例收口五种单模型；09 用集成学习（随机森林 / GBDT / XGBoost）把性能再推一步，随后补上 PCA 降维与核方法两个专题，最后用特征工程与综合项目接回测。本系列只负责“原理 + 案例”，模型上线等工程层列为后续计划。

**数据约定（跨电脑可复现）**：主线全部使用 sklearn 库内置数据或固定随机种子生成的模拟数据，零下载、零许可风险，任何电脑装上标准库（numpy / matplotlib / scikit-learn / xgboost）即可运行。例外有五处：综合案例（08、09）与工程案例（12）中的信用卡违约数据使用课程自带 UCI 官方数据集（CC BY 4.0）的 CSV 副本；13 章使用 GitHub 上 seaborn 官方数据仓库的鸢尾花数据（iris.csv）；14 章使用 GitHub 上 Coin Metrics 官方仓库的比特币数据（btc.csv）；15 章使用 GitHub 上 Fashion-MNIST 官方仓库的服装图像数据；16 章使用 GitHub 上 QuantConnect/Lean 官方仓库的美股日线样例数据；17 章使用 GitHub 上 justmarkham/pycon-2016-tutorial 仓库的垃圾短信数据（UCI SMS Spam Collection 整洁版）——均随 notebook 保存本地副本，同样离线可复现。

## 推荐学习顺序

1. **基础（01–02）**：线性回归建立回归评估与梯度下降；逻辑回归把线性模型扩展为概率分类，两类问题共用同一套“损失 → 梯度 → 更新”范式。
2. **经典分类（03–06）**：决策树、朴素贝叶斯、SVM、KNN 代表四种不同归纳偏置：划分规则、概率推断、间隔最大化、局部记忆。
3. **无监督与综合（07–08）**：KMeans 用无标签数据做分组；信用卡综合案例把五种单模型放进统一评估框架，比较交叉验证、阈值、校准与特征重要性。
4. **集成（09）**：随机森林（Bagging）与 GBDT/XGBoost（Boosting）是表格数据实战的主力，注意并行降方差与串行降偏差的区别。
5. **进阶专题（10–11）**：PCA 解决高维冗余、加速与可视化；核方法把 05 章 SVM 的核技巧推广到岭回归与 PCA。
6. **工程与综合（12–16）**：特征工程与调优补齐工程链路；13 章用鸢尾花把 16 个模型（3 无监督 + 9 分类 + 4 回归）放进同一尺度横评；14 章把骨架套到比特币时序上，以回归为主线并做「收益 vs 波动率」对照；15 章套到 Fashion-MNIST 图像上，补齐高维像素处理与错误分析；16 章完成量化 + ML 综合项目（截面因子 → 全模型横评 → 回测闭环）；17 章完成文本分类（TF-IDF → 全模型横评 → 垃圾词诊断），至此覆盖表格/时序/图像/金融/文本五大类数据形态；量化后续扩展（股票池、因子深水区、组合优化）可接独立的量化研究项目继续深化。

## 三套章节模板

**模板 A：模型章（01–11）**——模型原理 → NumPy 手写实现 → sklearn 调用与真实案例 → 章节小结与验证汇总。

**模板 B：工程流程章（12）**——流程框架总览 → 分步演示 → 全流程演练 → 小结；
不包含数学推导与手写实现，以贴近真实工程做法。

**模板 C：综合项目章（13–16）**——统一骨架：**数据 → 探索（无监督先上岗）→ 特征工程 → 全模型横评 → 诊断 → 结论**，
按数据形态换招式：

| 项目章 | 探索 | 特征工程 | 诊断 |
|---|---|---|---|
| 13 表格 | PCA / 核 PCA / KMeans | 标准化 + 衍生特征 | 决策边界 / 混淆矩阵 |
| 14 时序 | 波动率聚集 | 滞后 + 滚动特征 | 残差 + 对照实验（收益 vs 波动率） |
| 15 图像 | PCA 投影 / 图像重建 | 标准化 + PCA 降维 | 错误样本 + 降维对比 |
| 16 金融面板 | KMeans 市场状态 / PCA 市场因子 | 截面因子 + 横截面标准化 | **回测**：净值 / 回撤 / 月度热力图 |
| 17 文本 | 词频 / 词云 | TF-IDF（词 + 词组） | 错误短信 / 垃圾词诊断 |

**通用案例三件套**：独立测试集（或样本外验证）、指标表、可视化。
分类关注混淆矩阵与 ROC；回归关注真实值—预测值、残差与 R²；
聚类关注簇结构与参数选择（肘部法、轮廓系数）；集成与树模型关注特征重要性；
降维关注解释方差、重建误差与下游任务效果；量化关注 IC、AUC、净值与回撤。
